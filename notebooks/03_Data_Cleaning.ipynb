# 03 - Data Cleaning

This notebook cleans and prepares the data for modeling.

## Objectives:
- Remove duplicates
- Handle missing values
- Correct data types
- Merge datasets
- Save cleaned data

```python
import sys
sys.path.append('..')

import pandas as pd
import numpy as np

from src.data_loader import DataLoader
from src.preprocessing import DataPreprocessor
from src.utils import save_dataframe, reduce_memory_usage

import warnings
warnings.filterwarnings('ignore')
```

## 1. Load Raw Data

```python
loader = DataLoader(data_dir='../data/raw')
calendar_df = loader.load_calendar()
sales_df = loader.load_sales()
prices_df = loader.load_prices()

print("Data loaded successfully!")
print(f"Calendar: {calendar_df.shape}")
print(f"Sales: {sales_df.shape}")
print(f"Prices: {prices_df.shape}")
```

## 2. Clean Calendar Data

```python
preprocessor = DataPreprocessor()
calendar_clean = preprocessor.clean_calendar(calendar_df)

print("Calendar data cleaned!")
print(f"Shape: {calendar_clean.shape}")
print(f"\nMissing values after cleaning:")
print(calendar_clean.isnull().sum())

calendar_clean.head()
```

## 3. Clean Sales Data

```python
sales_clean = preprocessor.clean_sales(sales_df)

print(f"Sales data cleaned!")
print(f"Original shape: {sales_df.shape}")
print(f"Cleaned shape: {sales_clean.shape}")
print(f"Rows removed: {len(sales_df) - len(sales_clean)}")
```

## 4. Transform Sales to Long Format

```python
print("Transforming sales data to long format...")
sales_long = preprocessor.melt_sales_data(sales_clean)

print(f"\nSales data melted!")
print(f"Shape: {sales_long.shape}")
print(f"\nFirst few rows:")
sales_long.head()
```

## 5. Clean Prices Data

```python
prices_clean = preprocessor.clean_prices(prices_df)

print(f"Prices data cleaned!")
print(f"Original shape: {prices_df.shape}")
print(f"Cleaned shape: {prices_clean.shape}")
print(f"\nPrice statistics:")
print(prices_clean['sell_price'].describe())
```

## 6. Merge All Datasets

```python
print("Merging all datasets...")
merged_df = preprocessor.merge_datasets(sales_long, calendar_clean, prices_clean)

print(f"\nMerged data shape: {merged_df.shape}")
print(f"\nColumns: {list(merged_df.columns)}")
print(f"\nMissing values:")
print(merged_df.isnull().sum()[merged_df.isnull().sum() > 0])
```

## 7. Handle Missing Values

```python
print("Handling missing values...")
merged_df = preprocessor.handle_missing_values(merged_df, strategy='zero')

print(f"\nMissing values after handling:")
print(f"Total missing: {merged_df.isnull().sum().sum()}")
```

## 8. Optimize Memory Usage

```python
print("Optimizing memory usage...")
merged_df = reduce_memory_usage(merged_df)
```

## 9. Save Cleaned Data

```python
save_dataframe(merged_df, '../data/processed/cleaned_data.csv')

print("\nCleaned data saved successfully!")
print(f"Final shape: {merged_df.shape}")
print(f"Date range: {merged_df['date'].min()} to {merged_df['date'].max()}")
```

## Summary

Data cleaning completed:
- Removed duplicates
- Handled missing values
- Transformed to long format
- Merged all datasets
- Optimized memory usage
- Saved cleaned data
